```python
%%writefile ../app/app.py
"""APL Logistics: delivery performance dashboard (descriptive part)."""
import sys
from pathlib import Path
sys.path.append(str(Path(__file__).resolve().parents[1]))

import pandas as pd
import plotly.express as px
import streamlit as st

from src import config
from src.data import delivered_only
from src.kpis import group_summary

st.set_page_config(page_title="APL Logistics - Delivery Delays", layout="wide")


@st.cache_data
def load_data() -> pd.DataFrame:
    return delivered_only(pd.read_parquet(config.PROCESSED_PATH)).reset_index(drop=True)


d_all = load_data()

st.title("Delivery performance and delay analysis")
st.info(
    "Data note: in this dataset, lateness follows the chosen shipping mode very "
    "regularly (for example, First Class always takes 2 days against a 1-day promise). "
    "This looks like a simplified or generated dataset, so treat the patterns as "
    "properties of this file, not as proven real-world operating facts. "
    "Canceled shipments (7,754 orders) are excluded."
)

# ---------- Sidebar filters ----------
st.sidebar.header("Filters")


def pick(label, col):
    options = sorted(d_all[col].unique())
    return st.sidebar.multiselect(label, options, default=options)


modes = pick("Shipping mode", config.SHIP_MODE)
markets = pick("Market", config.MARKET)
segments = pick("Customer segment", config.SEGMENT)
regions = pick("Order region", config.REGION)

d = d_all[
    d_all[config.SHIP_MODE].isin(modes)
    & d_all[config.MARKET].isin(markets)
    & d_all[config.SEGMENT].isin(segments)
    & d_all[config.REGION].isin(regions)
]
if d.empty:
    st.warning("No orders match these filters.")
    st.stop()

# ---------- KPIs ----------
c1, c2, c3, c4 = st.columns(4)
c1.metric("Delivered orders", f"{len(d):,}")
c2.metric("Late rate", f"{d[config.TARGET].mean():.1%}")
c3.metric("On-time rate (early + on-time)", f"{(d[config.GAP] <= 0).mean():.1%}")
c4.metric("Average delay gap (days)", f"{d[config.GAP].mean():.2f}")

# ---------- Class mix and gap distribution ----------
left, right = st.columns(2)

mix = (d[config.DELIVERY_CLASS].value_counts(normalize=True)
         .reindex(["Early", "On-time", "Delayed"]).fillna(0).reset_index())
mix.columns = ["class", "share"]
fig = px.bar(mix, x="class", y="share", text=mix["share"].map("{:.1%}".format),
             title="Delivery outcome mix", color="class",
             color_discrete_map={"Early": "#55a868", "On-time": "#4c72b0", "Delayed": "#c44e52"})
fig.update_layout(showlegend=False, yaxis_tickformat=".0%", yaxis_title=None, xaxis_title=None)
left.plotly_chart(fig, use_container_width=True)

gap = d[config.GAP].value_counts().sort_index().reset_index()
gap.columns = ["delay_gap", "orders"]
fig = px.bar(gap, x="delay_gap", y="orders",
             title="Delay gap (real minus scheduled days)")
fig.update_xaxes(dtick=1)
right.plotly_chart(fig, use_container_width=True)

# ---------- Shipping mode ----------
st.subheader("By shipping mode")
m = group_summary(d, config.SHIP_MODE).sort_values("late_rate").reset_index()
left, right = st.columns([2, 3])
fig = px.bar(m, x="late_rate", y=config.SHIP_MODE, orientation="h",
             hover_data=["orders", "avg_gap"], title="Late rate by shipping mode")
fig.update_layout(xaxis_tickformat=".0%", xaxis_range=[0, 1])
left.plotly_chart(fig, use_container_width=True)

mode_class = pd.crosstab(d[config.SHIP_MODE], d[config.DELIVERY_CLASS], normalize="index")
mode_class = mode_class.reindex(columns=["Early", "On-time", "Delayed"], fill_value=0)
fig = px.bar(mode_class.reset_index().melt(id_vars=config.SHIP_MODE),
             x=config.SHIP_MODE, y="value", color=config.DELIVERY_CLASS,
             title="Outcome mix within each shipping mode",
             color_discrete_map={"Early": "#55a868", "On-time": "#4c72b0", "Delayed": "#c44e52"})
fig.update_layout(yaxis_tickformat=".0%", yaxis_title=None)
right.plotly_chart(fig, use_container_width=True)
st.dataframe(m.set_index(config.SHIP_MODE), use_container_width=True)

# ---------- Region ----------
st.subheader("By order region")
r = group_summary(d, config.REGION).sort_values("late_rate").reset_index()
fig = px.bar(r, x="late_rate", y=config.REGION, orientation="h",
             hover_data=["orders", "avg_gap"], title="Late rate by order region",
             height=620)
fig.add_vline(x=d[config.TARGET].mean(), line_dash="dash")
fig.update_layout(xaxis_tickformat=".0%")
st.plotly_chart(fig, use_container_width=True)
st.caption(
    "Regions differ by only a few percentage points, and most of that is explained by "
    "each region's shipping-mode mix. Small regions (hover to see order counts) are noisy."
)
```
